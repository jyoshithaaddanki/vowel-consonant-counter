```python
text = input("Enter your text: ").lower()

vowels = 0
consonants = 0

for char in text:
    if char.isalpha():
        if char in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("\n===== TEXT ANALYSIS =====")
print("Vowels:", vowels)
print("Consonants:", consonants)
```
